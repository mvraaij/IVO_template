````python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

# Data
x0, x1 = 0, 10
x = np.linspace(x0, x1, 10)
y = x**2


# range for dx
dx_min, dx_max, steps = 0.005, 1, 1000
dx_vals = np.linspace(dx_min, dx_max, steps)


def update(dx):
    # Model
    x_range = np.arange(x0, x1, dx)
    y_range = x_range**2
    # x_mid = (x_range[:-1]+ x_range[1:])/2
    # y_mid = x_mid**2
    x_links = x_range[:-1]
    y_links = x_links**2
    x_rechts = x_range[1:]
    y_rechts = x_rechts**2

    # Plots
    plt.figure(figsize=(15, 5))
    plt.subplot(1,2,1)
    plt.plot(x_range, y_range)
    plt.plot(x_links, y_links,'b.', markersize=5)
    plt.bar(x_links,y_links,width=dx,alpha=0.2,edgecolor='b', align='edge') # linker Riemannsom
    plt.ylabel('$x^2$')
    plt.xlabel('$x$')
    plt.title('Linkersom voor $x^2$')

    plt.subplot(1,2,2)
    plt.plot(x_range, y_range)
    plt.plot(x_rechts, y_rechts,'b.', markersize=5)
    plt.bar(x_rechts,y_rechts,width=-dx,alpha=0.2,edgecolor='b', align='edge') # rechter Riemannsom
    plt.ylabel('$x^2$')
    plt.xlabel('$x$')
    plt.title('Rechtersom voor $x^2$')

    #
    # plt.plot(x_range, y_range)
    # plt.plot(x_mid, y_mid,'b.', markersize=5)
    # plt.bar(x_mid,y_mid,width=dx,alpha=0.2,edgecolor='b') # Midpoint Riemannsum
    # plt.ylabel('$x^2$')
    # plt.xlabel('$x$')
    # plt.title('Riemannsom voor $x^2$')
    plt.show()

    print("L = ", np.sum(y_links*dx))
    print("R = ", np.sum(y_rechts*dx))

interact(
    update,
    dx=widgets.FloatSlider(min=dx_min, max=dx_max, step=(dx_max - dx_min) / steps, value=1, description='$\Delta x$')
)
````

De widget hierboven laat zien wat de linker- en rechtersommen zijn van de functie $x^2$ voor een variabele waarde van $\Delta x$. Daarnaast worden ook de waarden van deze sommen berekend.
Wanneer $\Delta x$ kleiner wordt, is te zien dat de waardes van de Riemannsommen steeds minder verschillen.
